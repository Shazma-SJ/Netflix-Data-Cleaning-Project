```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("netflix_cleaned.csv")
print(df)
```

```python
###   to check correct file is opened

print(df.shape)
print(df.head())
print(df["Release Year"].dtype)
```

```python
###   Organize content by release year and calculate yearly content

yearly_content=df["Release Year"].value_counts().sort_index()
print(yearly_content)
```

```python
###   Growth and decline trends   ###

yearly_growth=yearly_content.pct_change()*100
print(yearly_growth.round(2))
print(df["Date Added"].max())
```

```python
import matplotlib.ticker as ticker

###   Trend Visualization   ###
plt.figure(figsize=(12,8))
plt.plot(yearly_content.index, yearly_content.values, marker="o", markersize=4)
plt.title("Netflix Content Analysis by release year", size=20)
plt.xlabel("Release year", size=16)
plt.ylabel("Number of titles", size=16)

###   X-Axis: with gap of 10 years  ###
plt.gca().xaxis.set_major_locator(ticker.MultipleLocator(10))

###   Y-axis: with gap of 100 titles  ###
plt.gca().yaxis.set_major_locator(ticker.MultipleLocator(100))
plt.grid(True)
plt.show()
```

## Task 4: Content Trend Analysis by Release Year — Summary

| Observation | Detail |
|---|---|
| Growth Period | Minimal content before 2000, sharp rise from 2010 onward |
| Peak Year | ~2018-2019, with over 1,100 titles released |
| Recent Decline | Drop to ~600 titles by 2021, likely reflecting incomplete/lagging recent data |

### Conclusion
Netflix's content library shows minimal growth before the 2000s, followed by an exponential increase in titles released from 2010 onward. Content peaked around 2018-2019. The apparent decline after 2019 likely reflects incomplete data for the most recent years rather than an actual reduction in content output.
